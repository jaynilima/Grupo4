# Coleta de preços ajustados pelo Yahoo Finance

Nesta etapa, utilizamos o universo líquido construído a partir da COTAHIST da B3 para baixar os preços ajustados dos ativos pelo Yahoo Finance.

A COTAHIST foi usada para definir quais ações eram líquidas em cada janela histórica. Já o Yahoo Finance será usado para obter preços ajustados por eventos como dividendos, juros sobre capital próprio, desdobramentos e grupamentos.

```python
from pathlib import Path

import pandas as pd
import yfinance as yf
```

```python
PASTA_PROJETO = Path.cwd().parent
PASTA_DADOS_TRATADOS = PASTA_PROJETO / "dados_tratados"

universo_liquido = pd.read_csv(
    PASTA_DADOS_TRATADOS / "universo_liquido_por_janela.csv",
    parse_dates=[
        "inicio_formacao",
        "fim_formacao",
        "inicio_teste",
        "fim_teste"
    ]
)

universo_liquido.head()
```

```python
ativos_b3 = sorted(universo_liquido["ativo"].unique())

print(f"Quantidade de ativos únicos: {len(ativos_b3)}")
ativos_b3[:20]
```

```python
tickers_yahoo = [ativo + ".SA" for ativo in ativos_b3]

tickers_yahoo[:20]
```

```python
dados_yahoo = yf.download(
    tickers=tickers_yahoo,
    start="2010-01-01",
    end="2026-01-01",
    interval="1d",
    auto_adjust=True,
    group_by="column",
    threads=True,
    progress=True
)
```

```python
precos_ajustados = dados_yahoo["Close"]

precos_ajustados.head()
```

```python
precos_ajustados.columns = precos_ajustados.columns.str.replace(".SA", "", regex=False)

precos_ajustados.head()
```

```python
print(f"Matriz de preços ajustados: {precos_ajustados.shape[0]} datas e {precos_ajustados.shape[1]} ativos")
print(f"Data inicial: {precos_ajustados.index.min()}")
print(f"Data final: {precos_ajustados.index.max()}")

precos_ajustados.iloc[:5, :10]
```

### 1. Conferência dos dados faltantes nos preços ajustados

Após baixar os preços ajustados pelo Yahoo Finance, é necessário verificar a quantidade de dados faltantes por ativo. Essa etapa ajuda a identificar quais tickers possuem histórico incompleto ou não foram encontrados corretamente na base do Yahoo.

A presença de valores ausentes é esperada, pois nem todos os ativos existiam desde 2010 ou possuem histórico completo disponível no Yahoo Finance.

```python
percentual_dados_yahoo = precos_ajustados.notna().mean().sort_values(ascending=False)

percentual_dados_yahoo.head(20)
```
